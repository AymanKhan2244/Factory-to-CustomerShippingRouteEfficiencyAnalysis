## Geographic Bottleneck

```python
## state  level lead time analysis 

state_leadtime = df.groupby(
    'State/Province'
).agg(
    Avg_Lead_Time=('Lead_time','mean'),
    Total_Shipments=('Order ID','count')
).reset_index()

state_leadtime = state_leadtime.sort_values(
    by='Avg_Lead_Time',
    ascending=False
)

print(state_leadtime.head(10))
```

```python
## visualizations 

import matplotlib.pyplot as plt

top_bottlenecks = state_leadtime.head(10)

plt.figure(figsize=(12,6))

plt.barh(
    top_bottlenecks['State/Province'],
    top_bottlenecks['Avg_Lead_Time']
)

plt.xlabel("Average Lead Time")
plt.ylabel("State")
plt.title("Top 10 Bottleneck States")

plt.gca().invert_yaxis()

plt.show()
```

```python
region_analysis = df.groupby(
    'Region'
).agg(
    Avg_Lead_Time=('Lead_time','mean'),
    Total_Shipments=('Order ID','count')
).reset_index()

print(region_analysis)
```

```python
## HIGH VLOULME AND POOR REGIONS 

region_analysis.sort_values(
    by=['Total_Shipments','Avg_Lead_Time'],
    ascending=[False,False]
)
```

```python
state_congestion = df.groupby(
    'State/Province'
).agg(
    Total_Shipments=('Order ID','count'),
    Avg_Lead_Time=('Lead_time','mean')
).reset_index()

state_congestion = state_congestion.sort_values(
    by=['Total_Shipments','Avg_Lead_Time'],
    ascending=[False,False]
)

print(state_congestion.head(10))
```

```python
state_congestion['Congestion_Score'] = (
    state_congestion['Total_Shipments']
    * state_congestion['Avg_Lead_Time']
)
```

```python
## Detect congestion-prone states or regions

state_congestion = state_congestion.sort_values(
    by='Congestion_Score',
    ascending=False
)

print(state_congestion.head(10))
```

## DELAY FREQUENCY

```python
df['Delayed'] = df['Lead_time'].apply(
    lambda x: 1 if x > 5 else 0
)
```

```python
delay_frequency = df.groupby(
    'Route'
)['Delayed'].mean() * 100

print(delay_frequency.head())
```

## EFFIECIENCY SCORE

```python
max_lead = df["Lead_time"].max()
```

## KPI TABLE

```python
kpi_table = df.groupby('Route').agg(

    Average_Lead_Time=('Lead_time', 'mean'),

    Route_Volume=('Order ID', 'count'),

    Delay_Frequency=('Delayed', 'mean'),

    Efficiency_Score=('Efficiency Score', 'mean')

).reset_index()
```

```python
kpi_table['Delay_Frequency'] = (
    kpi_table['Delay_Frequency'] * 100
)
```
