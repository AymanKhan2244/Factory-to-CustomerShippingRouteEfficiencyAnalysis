```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    dayfirst=True,
    errors='coerce'
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    dayfirst=True,
    errors='coerce'
)
```

```python
df["Lead_time"] = (df["Ship Date"] - df["Order Date"]).dt.days
```
